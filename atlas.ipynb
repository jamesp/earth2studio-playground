```python
pkg = Atlas.load_default_package()
model = Atlas.load_model(pkg)
```

```python
data = ARCO()
io = ZarrBackend("2023-01-05-atlas-arco.zarr", backend_kwargs={"overwrite": True})
nsteps = 1
fcast = run.deterministic(["2023-01-05"], nsteps, model, data, io)
```

```python
import xarray as xr
ds = xr.open_dataset("2023-01-05-atlas-arco.zarr")
ds
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

fig, ax = plt.subplots(subplot_kw={"projection": ccrs.Robinson()})

ds['msl'].isel(time=0, lead_time=1).plot.pcolormesh(
    ax=ax, 
    transform=ccrs.PlateCarree(), 
    cmap='viridis',
    add_colorbar=False
    )
ax.coastlines()
```
